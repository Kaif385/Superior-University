```python
MAX_X = 3
MAX_Y = 4
TARGET = 2

def get_neighbors(state):
    x, y = state
    neighbors = []
    if x < MAX_X:
        neighbors.append(((MAX_X, y), "Fill X"))
    if y < MAX_Y:
        neighbors.append(((x, MAX_Y), "Fill Y"))
    if x > 0:
        neighbors.append(((0, y), "Empty X"))
    if y > 0:
        neighbors.append(((x, 0), "Empty Y"))
    if x > 0 and y < MAX_Y:
        transfer = min(x, MAX_Y - y)
        neighbors.append(((x - transfer, y + transfer), f"Pour X->Y ({transfer})"))
    if y > 0 and x < MAX_X:
        transfer = min(y, MAX_X - x)
        neighbors.append(((x + transfer, y - transfer), f"Pour Y->X ({transfer})"))
    return neighbors

def dfs(start=(0, 0)):
    stack = [(start, [])]
    visited = set([start])
    while stack:
        state, path = stack.pop()
        x, y = state
        if x == TARGET or y == TARGET:
            return path + [("Goal reached", state)]
        for next_state, action in get_neighbors(state):
            if next_state not in visited:
                visited.add(next_state)
                stack.append((next_state, path + [(action, next_state)]))
    return None

solution = dfs()
if solution:
    print("DFS Solution Path:\n")
    for i, (action, state) in enumerate(solution, 1):
        print(f"Step {i}: {action} -> {state}")
else:
    print("No solution found.")
```
