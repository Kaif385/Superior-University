## N-QUEEN PROBLEM (DYNAMIC)

```python
def issafe(grid, x, y, n):
    for row in range(x):
        if grid[row][y] == 1:
            return False

    row, col = x - 1, y - 1
    while row >= 0 and col >= 0:
        if grid[row][col] == 1:
            return False
        row -= 1
        col -= 1

    row, col = x - 1, y + 1
    while row >= 0 and col < n:
        if grid[row][col] == 1:
            return False
        row -= 1
        col += 1

    return True
```

```python
def nqueen(grid, x, n, solutions):
    if x == n:
        sol = []
        for r in grid:
            row = "".join("Q" if c == 1 else "." for c in r)
            sol.append(row)
        solutions.append(sol)
        return

    for y in range(n):
        if issafe(grid, x, y, n):
            grid[x][y] = 1     
            nqueen(grid, x + 1, n, solutions)
            grid[x][y] = 0  
```

```python
def solve_nqueen_dynamic():
    n = int(input("Enter the value of N (board size): "))
    grid = [[0 for _ in range(n)] for _ in range(n)]
    solutions = []

    nqueen(grid, 0, n, solutions)

    print(f"\nTotal solutions for {n}-Queens = {len(solutions)}\n")
    for idx, sol in enumerate(solutions, 1):
        print(f"Solution {idx}:")
        for row in sol:
            print(row)
        print()
```

```python
solve_nqueen_dynamic()
```
