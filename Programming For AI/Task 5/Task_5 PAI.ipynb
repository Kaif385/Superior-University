```python
import cv2
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt
```

## Read Image

```python
img = cv2.imread(r"Cat03.jpg")
```

## Convert Color

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

## Resize Image

```python
resized = cv2.resize(img, (400, 300))
```

## Rotate Image

```python
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
```

## Crop Image

```python
cropped = img[50:200, 100:300]
```

## Draw shapes and text

```python
draw = img.copy()
cv2.rectangle(draw, (50, 50), (200, 200), (0, 255, 0), 3)
cv2.circle(draw, (300, 300), 50, (255, 0, 0), -1)
cv2.putText(draw, "OpenCV", (60, 45), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)
```

## Blur % Edge Detection

```python
blur = cv2.GaussianBlur(img, (7,7), 0)
edges = cv2.Canny(img, 100, 200)
```

## Save Output images

```python
cv2.imwrite("gray_image.jpg", gray)
cv2.imwrite("resized_image.jpg", resized)
cv2.imwrite("rotated_image.jpg", rotated)
cv2.imwrite("cropped_image.jpg", cropped)
cv2.imwrite("draw_image.jpg", draw)
cv2.imwrite("blur_image.jpg", blur)
cv2.imwrite("edges_image.jpg", edges)

print("✅ All main OpenCV functions implemented successfully!")
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title("Original")
plt.subplot(2,2,2); plt.imshow(gray, cmap='gray'); plt.title("Grayscale")
plt.subplot(2,2,3); plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB)); plt.title("Blurred")
plt.subplot(2,2,4); plt.imshow(edges, cmap='gray'); plt.title("Edges")
plt.show()
```
