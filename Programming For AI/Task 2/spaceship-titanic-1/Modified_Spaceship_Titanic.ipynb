```python
def fillNaMean(KF, categorical_cols):
    for col in categorical_cols:
        KF[col] = KF[col].fillna(KF[col].mode()[0])
categorical_columns = [
    'HomePlanet','CryoSleep','Cabin','Destination',
    'Age','VIP','RoomService','FoodCourt',
    'ShoppingMall','Spa','VRDeck','Name'
]
```

```python
def changeFloatToInt64(cols):
    for i in cols:
        KF[i] = KF[i].replace([np.inf, -np.inf], np.nan)

        KF[i] = KF[i].fillna(0)

        KF[i] = KF[i].astype('int64')

columns = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
changeFloatToInt64(columns)
```

```python
KF.info()
```

```python
def encodeCols(cols):
    for i in cols:
        temp = pd.DataFrame({i:KF[i].unique()})
        data_LE = LabelEncoder()
        data_LE.fit(np.ravel(temp))
        KF[i] = data_LE.transform(KF[i])

columns = ['PassengerId','HomePlanet','Cabin','Destination','Name','CryoSleep','VIP']
encodeCols(columns)
```

```python
KF.info()
```

```python
KF.drop('PassengerId', axis=1, inplace=True)
KF.drop('Name', axis=1, inplace=True)
KF.drop('RoomService', axis=1, inplace=True)
```

```python
KF.info()
```

```python
KF['Transported'] = KF['Transported'].astype(int)
```

```python
X = KF.drop(columns=['Transported'])
y = KF['Transported']
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
model = RandomForestClassifier()
model.fit(X, y)
```
